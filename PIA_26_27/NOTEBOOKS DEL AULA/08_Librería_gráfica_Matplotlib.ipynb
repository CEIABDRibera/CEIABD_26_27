## Introducción   

El paquete `matplotlib` es una extensa biblioteca de funciones para generar gráficos 2D y 3D.

Algunas de sus características son:

- Curva de aprendizaje suave: ofrece funcionalidades simples para usuarios ocasionales.

- Todo tipo de ajustes programables de los elementos de una ventana gráfica.

- Gran variedad de formatos de exportación de las figuras.

`Matplotlib` está construido sobre el paquete `Numpy` y trabaja de forma natural con los vectores y matrices (arrays) a ella asociados. 
Inicialmente, vamos a trabajar con las listas nativas de Python (que serán transformadas internamente a numpy arrays en las funciones del paquete Matlotlib). 
Restringiremos el uso a listas formadas por valores del mismo tipo de dato.

Aunque no es la forma usual, utilizar en este momento listas nos permite:

- visualizar desde ya los resultados de algunos de nuestros ejercicios.

- nos obliga a practicar con el manejo de las estructuras de control; no olvidemos que nuestro objetivo es aprender a programar. De la otra forma, muchas operaciones de bajo nivel quedan ocultas detrás de las poderosas herramientas que nos brinda el paquete Numpy.

Dentro del paquete matplotlib destaca el módulo **pyplot**. Este módulo oculta muchas de las funcionalidades de bajo nivel de la biblioteca, permitiendo el uso de sencillas funciones para los elementos gráficos más habituales, tales como creación de figuras, trazado de líneas, visualización de imágenes, inserción de texto, etc.   

El módulo `pyplot` emula el entorno de programación gráfica de **MATLAB**, herramienta software estadístico muy popular.

Para comenzar usando el módulo `pyplot` de **Matplotlib** en un programa Python la forma estándar es:

```python
import matplotlib.pyplot as plt
```

## Gráficos básicos    

### Función `plot()`   

Mediante la función `.plot()` suministramos una lista para su trazado.

```python
lista = [6,2,5,6,8,1,3,6,7]
plt.plot(lista)
```

Por defecto, `.plot(lista)` une con una línea la secuencia de puntos de coordenadas, siendo los índices válidos de la lista desde *(i, lista[i])*, siendo *i* los índices desde 0 hasta el último índice de la lista (*8* en este caso). En nuestro ejemplo, los puntos son *{(0,6),(1,2),(2,5),(3,6),(4,8),(5,1),(6,3),(7,6),(8,7)}*

En este primer ejemplo, la figura donde reside la línea trazada *se crea de forma automática para contener el dibujo*. No solo eso, en ausencia de indicaciones, se usan diferentes opciones por defecto, como tamaño de la figura, color y grosor de la línea, etc.

Si proporcionamos dos listas, `.plot(lista_x, lista_y)`, se une con una línea la secuencia de puntos de coordenadas `(lista_x[i], lista_y[i])`, siendo *i* los índices válidos de las listas desde 0. Es decir, la primera lista corresponde a las abscisas y la segunda a las ordenadas. Obviamente, ambas listas tendrán el mismo número de elementos. 

```python
plt.plot([1.1, -3.23, 5.3, 7.34, 6.6, 0.1, 2.123, 4.17],
         [6, 2, 5.12, 6.25, 8.2, 1, -3.35, 6.3])
```

### Función `show()`   

En los dos ejemplos anteriores el kernel de *IPython* ha mostrado las figuras sin necesidad de usar ninguna función añadida. En realidad, tras la ejecución de una celda, *se ha invocado internamente* a la función `.show()`, que es la que de forma efectiva muestra la figura.

Se usa la llamada a la función `.show()` para controlar el momento en que se muestra la figura.

```python
plt.plot([1.1, -3.23, 5.3, 7.34, 6.6, 0.1, 2.123, 4.17],
         [6, 2, 5.12, 6.25, 8.2, 1, -3.35, 6.3])
plt.show()
```

### Colores y formatos de línea.   

Además de los parámetros correspondientes a las listas, la función `.plot()` admite una cadena con una serie de caracteres dotados de significado como tercer argumento opcional. El significado de los caracteres está *prestado* de **MATLAB**: sirven para indicar el color, el tipo de línea y si el punto tiene una marca asociada.    

El formato por defecto es `'b-'`, una línea sólida azul (blue). Basta utilizar `'r^--'` para que la línea sea roja `(r)` discontinua `(--)` con marcas triangulares `(^)` en los puntos o `'go'` para dibujar los puntos circulares `(o)` de forma aislada en color verde `(g)`.

```python
plt.plot(range(10), 'r^--')
plt.show()
```

```python
plt.plot(range(10), 'go')
plt.show()
```

### Función `scatter()`    

Mediante la función .scatter() podemos dibujar los puntos *(lista_x[i], lista_y[i])* sin unir por líneas, apareciendo los puntos de forma dispersa.

Sin ningún tipo de argumento opcional es equivalente a usar la función `.plot()` con el argumento `'o'`.


```python
plt.scatter([1.1, -3.23, 5.3, 7.34, 6.6, 0.1, 2.123, 4.17],
            [6, 2, 5.12, 6.25, 8.2, 1, -3.35, 6.3])
plt.show()
```

Sin embargo, las posibilidades que ofrece a mayores respecto a `.plot()` son muy diversas. Por ejemplo, con el parámetro opcional` s `podemos incluir una lista con los radios deseados de los puntos y con el parámetro opcional`c`, sus colores. Eso entre otras muchas opciones.

```python
lista_x = []
lista_y = []

radio = 1
radios = []
rojo_verde_azul = [(1, 0, 0), (0, 1, 0), (0, 0, 1)]  # Colores R, G y B
colores = []
for i in range(3): # i toma los valores 0, 1 y 2, que corresponden a las coordenadas x de los puntos que se van a crear
    color = rojo_verde_azul[i]
    for j in range(3): # En este bucle se crean 3 puntos con el mismo color, aumentando el radio de cada uno el doble del anterior
        lista_x.append(i)
        lista_y.append(j)
        radios.append(radio)
        colores.append(color)
        radio *= 2

plt.scatter(lista_x, lista_y, s=radios, c=colores)
plt.show()
print("Radios de los puntos:", radios)
```

### Graficando funciones de una sola variable

Un uso habitual de las herramientas gráficas es dibujar una función $y=f(x)$ para una colección de puntos en un intervalo de valores.

Por ejemplo, supongamos que deseamos dibujar la función $y=cos²(x)$ utilizando 100 puntos equidistantes en el intervalo $-2\pi\leq x \leq 2\pi$.

```python
import matplotlib.pyplot as plt
import math
num_puntos = 100
x_min = -2*math.pi
x_max = -x_min

print("X inicial:", x_min)
print("X final:", x_max)        

# Hay (num_puntos - 1) intervalos incluyendo x_min y x_max
incremento = (x_max - x_min)/(num_puntos - 1)

lista_x = [0]*num_puntos  # Rellenamos de 0's ambas listas. Así evitamos usar el método append().
lista_y = [0]*num_puntos
for i in range(num_puntos): #bucle que recorre los índices de las listas, que van de 0 a num_puntos-1(99)
    x = x_min + i*incremento  # x va tomando los num_puntos equidistantes entre x_min y x_max
    lista_x[i] = x  # valor para la coordenada x del punto i
    lista_y[i] = math.cos(x)**2  # imagen resultado de aplicar la función cos^2 a x, que será la coordenada y del punto i

plt.plot(lista_x, lista_y)
plt.show()
```

#### Dibujando varias gráficas superpuestas
Es muy sencillo dibujar varias funciones superpuestas; tan solo es necesario invocar a la función `.plot()` de manera consecutiva, incluso con listas de tamaños diferentes.    

En el ejemplo, se añade una gráfica de la función $y=sin(x)cos(x)$ utilizando 50 puntos en el intervalo $-3 \leq x \leq 3$.

Como se puede observar, *Matplotlib* configura el gráfico de forma automática para adaptarse a los distintos rangos y eligiendo colores diferentes.

```python
num_puntos = 50
x_min = -3
x_max = -x_min

print("X inicial:", x_min)
print("X final:", x_max)

incremento = (x_max-x_min)/(num_puntos - 1)

print("Incremento:", incremento)

lista_x2 = [0]*num_puntos
lista_y2 = [0]*num_puntos
for i in range(num_puntos):
    x = x_min + i*incremento
    lista_x2[i] = x
    lista_y2[i] = math.sin(x)*math.cos(x)

plt.plot(lista_x, lista_y)   # Esta línea dibuja la función cos^2(x) que ya habíamos dibujado en el bloque anterior.
plt.plot(lista_x2, lista_y2) # Esta línea dibuja la función sin(x)*cos(x) que acabamos de calcular.
plt.show()
```

### Primeros pasos: construyendo un gráfico pieza a pieza.   

Para este ejemplo, tomaremos información de un dataset deportivo con información estadística relativo a ligas europeas (*football_stats.csv*).

```python
import pandas as pd

df_football = pd.read_csv('./datasets/football_stats.csv')
df_football.head()
```

A continuación, realizamos una selección/filtrado de los datos para quedarnos con la información goleadora de Messi y Ronaldo entre 2014 y 2019.

```python
# Filtrando datos del dataset entre 2014 y 2019
messi_goals = df_football[df_football['player_name'] == 'Lionel Messi'].query('year != 2020')
ronaldo_goals = df_football[df_football['player_name'] == 'Cristiano Ronaldo'].query('year != 2020')
```

Seguidamente, graficamos la información dentro de nuestra `figura`.   

`fig` nos permite manipular la figura completa, mientras que `ax` nos permite manipular los ejes de la figura. En este caso, como solo tenemos un gráfico, `fig` y `ax` son únicos. Si tuviéramos varios gráficos en la misma figura, `fig` seguiría siendo único, pero habría varios `ax`.   

`ax.plot(X, Y)` es la forma de dibujar en el objeto `ax`, que es un eje de coordenadas. En nuestro ejemplo, tenemos dos líneas de `ax.plot(X, Y)`: una nos traza la información de Messi y la otra la información de Ronaldo, ambas dentro de la misma figura de gráfico único.

```python
fig, ax = plt.subplots() 
ax.plot(messi_goals.year, messi_goals.goals) 
ax.plot(ronaldo_goals.year, ronaldo_goals.goals)
# plt.savefig('1.jpg')
plt.show()
```

## Personalizando un plot

Hay una serie de argumentos que permiten personalizar cómo se ven los datos del plot, entre los que están los siguientes:

- `marker='o'` → añade marcadores al plot y también es posible indicar el tipo de marcador que quieres. Aqui están las [opciones de marcadores](https://matplotlib.org/stable/api/markers_api.html) que hay.
- `linestyle='--'` → cambia el tipo de línea. [Aquí](https://matplotlib.org/stable/gallery/lines_bars_and_markers/linestyles.html) se encuentran los distintos tipos de línea. Incluso si se usa 'None', simplemente ya no habrá líneas y quedarán solo los marcadores.
- `color='b'` → cambia el color de los datos. [Aquí](https://matplotlib.org/stable/users/explain/colors/colors.html) se muestra cómo poner cualquier color e incluso dejarlos transparentes.
- `label='name'` → pone una etiqueta al plot (muy útil cuando es necesario poner varias gráficas en un solo plot).


### Título y etiquetas en ejes.   

Para un gráfico sencillo, en el que no hemos especificado ni figura ni subplots ni ejes, simplemente graficamos datos, como la función usada anteriormente $cos^2(x)$, podemos personalizar el título y las etiquetas de los ejes usando las siguientes funciones:   

- La función `.title() `podemos añadir un **título** al gráfico.    
- Las funciones `.xlabel()` e `.ylabel()` añadimos **etiquetas a los ejes de coordenadas**.

```python
plt.plot(lista_x, lista_y)
plt.title('Función $cos^2(x)$')
plt.xlabel('$x$')
plt.ylabel('$cos^2(x)$')  # el simbolo $ indica que lo que hay entre ellos es una expresión matemática
plt.show()
```

Si partimos de un gráfico en el que hemos definido la figura, ejes, subplots,..., lo más conveniente es utilizar los métodos específicos que actúan sobre dichos objetos. De esta forma, para el objeto `Axes` se dispone de los siguientes métodos:   

- `.set_xlabel('Name')` → coloca nombre al eje de las X.
- `.set_ylabel('Name') `→ coloca nombre al eje de las Y.
- `.set_title('Name')` → coloca el título de tu plot.
- `.set(xlabel='Name') `→ también se puede usar únicamente `set` y pasar como parámetros todo lo anterior.
- `.legend() `→ si se han colocado etiquetas, legend las mostrará.    

Todo este tipo de configuración debe ir siempre antes de ejecutar `plt.show()`

```python
fig, ax = plt.subplots()
ax.plot(messi_goals.year, messi_goals.goals,
        marker='o', color='b', linestyle='--', label='Lionel Messi')
ax.plot(ronaldo_goals.year, ronaldo_goals.goals,
        marker='s', color='r', linestyle='dotted', label='Cristiano Ronaldo')

ax.set(xlabel='Años', ylabel='Goles', title='Goles: Messi vs. Cristiano (2014 - 2019)')
ax.legend()
# plt.savefig('2.jpg')
plt.show()
```

### Trabajando con `Subplots`

Si queremos representar múltiples gráficos, como ya hemos visto, es posible superponer información en un único gráfico. Esto es correcto si no se acumula un exceso de información que provoque ilegibilidad del gráfico, como sucede en el ejemplo siguiente:

```python
cavani_goals = df_football[df_football['player_name'] == 'Edinson Cavani'].query('year != 2020')
```

```python
fig, ax = plt.subplots()
ax.plot(messi_goals.year, messi_goals.goals,
        marker='o', color='b', linestyle='--', label='Lionel Messi')
ax.plot(messi_goals.year, messi_goals.npg,
        marker='.', color='b', linestyle='dotted')
ax.plot(ronaldo_goals.year, ronaldo_goals.goals,
        marker='s', color='r', linestyle='--', label='Cristiano Ronaldo')
ax.plot(ronaldo_goals.year, ronaldo_goals.npg,
        marker='.', color='r', linestyle='dotted')
ax.plot(cavani_goals.year, cavani_goals.goals,
        marker='D', color='m', linestyle='--', label='Edinson Cavani')
ax.plot(cavani_goals.year, cavani_goals.npg,
        marker='.', color='m', linestyle='dotted')

ax.set(xlabel='Años', ylabel='Goles', title='Goles y goles sin penalti')
ax.legend()
# plt.savefig('3.jpg')
plt.show()
```

El gráfico anterior empieza a resultar ilegible y apenas se puede distinguir la información de los distintos jugadores y de los distintos tipos de goles.   

Para resolver esto es aconsejable utilizar los **subplots**, que permiten, dentro de la misma figura, separar en distintos gráficos la información, haciendo que sea más legible y entendible para el usuario.    

Para construir una figura que muestre por separado los seis gráficos que se necesitan para mostrar la información anterior, se debe hacer de la siguiente manera:

```python
fig, ax = plt.subplots(3, 2) # en ax especificamos que queremos 3 filas y 2 columnas de subplots
# plt.savefig('4.jpg')
plt.show()
ax.shape
```

Ahora, sobre esta estructura, se construye cada uno de los subgráficos y se dibuja la gráfica correspondiente. El resultado que obtenemos nos facilita la visualización y la comprensión de la información representada.

```python
fig, ax = plt.subplots(3, 2, sharey=True, figsize=[6, 10])

ax[0,0].plot(messi_goals.year, messi_goals.goals,
        marker='o', color='b', linestyle='--', label='Goles totales')
ax[0,1].plot(messi_goals.year, messi_goals.npg, marker='.',
        color='b', linestyle='dotted', label='Goles de no penalti')
ax[1,0].plot(ronaldo_goals.year, ronaldo_goals.goals,
        marker='s', color='r', linestyle='--', label='Goles totales')
ax[1,1].plot(ronaldo_goals.year, ronaldo_goals.npg, marker='.',
        color='r', linestyle='dotted', label='Goles de no penalti')
ax[2,0].plot(cavani_goals.year, cavani_goals.goals,
        marker='D', color='m', linestyle='--', label='Goles totales')
ax[2,1].plot(cavani_goals.year, cavani_goals.npg, marker='.',
        color='m', linestyle='dotted', label='Goles de no penalti')

ax[2,0].set_xlabel('Años')
ax[2,1].set_xlabel('Años')
ax[0,0].set_ylabel('Lionel Messi')
ax[1,0].set_ylabel('Cristiano Ronaldo')
ax[2,0].set_ylabel('Edinson Cavani')
ax[0,0].set_title('Goles totales')
ax[0,1].set_title('Goles de no penalti')
ax[0,0].legend()
ax[0,1].legend()
ax[1,0].legend()
ax[1,1].legend()
ax[2,0].legend()
ax[2,1].legend()
# plt.savefig('5.jpg')
plt.show()
```

## Tipos básicos de gráficas.     

### Gráficas de barras.   

##### Características principales

• *Eje X (horizontal)*: Muestra las categorías, grupos o períodos de tiempo.    
• *Eje Y (vertical)*: Muestra la escala numérica con las frecuencias o cantidades.    
• *Barras*: Rectángulos de igual ancho cuya altura o longitud indica el valor de cada categoría.     
• *Título y leyenda*: Indican el tema del gráfico y explican los datos o series representadas.     

##### Uso    
El uso principal de un gráfico de barras es *representar un dato sintetizado en forma de medidas* para que el usuario pueda comparar fácilmente diferentes valores y sacar una conclusión rápidamente sobre los datos descritos.    



Para crear un gráfico de barras se puede hacer utilizando el método `plt.bar()` siendo *el primer argumento la variable del eje x*  y *el parámetro `height` el eje y*.  A continuación, es necesario asegurarnos de llamar a `plt.show()` para mostrar el gráfico.   

En el siguiente ejemplo, se trabaja con la información relativa a las tarjetas rojas de los distintos equipos:

```python
# Datos de tarjetas rojas por equipo
top_red_cards = df_football.groupby('team_name')['red_cards'].sum() \
                .sort_values(ascending=False).head(10)
```

```python

# Plot
plt.style.use('ggplot') # el parametro 'ggplot' es un estilo de gráficos que imita el estilo de la librería ggplot2 de R
plt.bar(top_red_cards.index, top_red_cards)
plt.xticks(rotation=90)
plt.title('Tarjetas rojas acumuladas')
plt.ylabel('Tarjetas rojas')
plt.show()
```

Podemos tener el mismo tipo de gráfico pero "trasponiendo" los ejes, de forma que nos quede:

```python
plt.barh(top_red_cards.index, top_red_cards)
plt.title('Tarjetas rojas acumuladas')
plt.xlabel('Tarjetas rojas')
plt.show()
```

Otro tipo específico de gráficos de barras es el de *barras apiladas*. Este tipo de gráfico permite de forma visualmente intuitiva representar y comparar datos categóricos.    

A diferencia de un gráfico de barras regular, el de *barras apiladas* muestra múltiples variables para cada categoría, con cada segmento apilado encima de los demás. Esto permite tener una visualización clara tanto del total como de la composición de cada categoría.     

La altura de cada segmento corresponde a su valor, lo que facilita la identificación de patrones, tendencias y proporciones relativas dentro de los datos.    

Los gráficos de barras apiladas son particularmente útiles para ***mostrar la relación entre diferentes componentes dentro de un conjunto de datos*** completo.

```python
# Filtramos los datos del dataset entre 2014 y 2019
df_football['goals_assists'] = df_football.goals + df_football.assists
top_players = df_football.groupby('player_name')[['goals', 'assists', 'goals_assists']].sum() \
                .sort_values('goals_assists', ascending=False).head(10)
```

```python
# Graficamos
plt.bar(top_players.index, top_players.goals, label='Goles')
#el parámetro `bottom` indica que las barras de asistencias se dibujarán encima de las barras de goles
plt.bar(top_players.index, top_players.assists, label='Asistencias', bottom=top_players.goals) 
plt.xticks(rotation=90)
plt.title('Goles y asistencias de los Top Players')
plt.ylabel('Goles y asistencias')
plt.legend()
plt.show()
```

### Histograma.    

Un **histograma** es una representación gráfica de la distribución de un conjunto de datos *cuantitativos* y *continuos* (como peso, altura, tiempo o temperatura) mediante barras.


##### Características principales

• *Eje horizontal (X)*: Representa los intervalos o clases (rangos de valores en los que se agrupan los datos).    
• *Eje vertical (Y)*: Representa la frecuencia (cuántas veces aparecen los datos en cada intervalo).     
• *Barras pegadas*: A diferencia de un gráfico de barras tradicional, en el histograma las barras van unidas porque representan una escala numérica continua sin espacios entre rangos.    
• *Superficie proporcional*: El área de cada barra es proporcional a la frecuencia de los valores que contiene.    


##### Uso

• *Visualizar la forma de la distribución*: Permite ver si los datos están concentrados en el centro, sesgados hacia un lado (asimetría) o si tienen dos picos.    
• *Identificar dispersión y valores atípicos*: Ayuda a detectar si hay mucha variabilidad o datos extremos alejados del resto.     
• *Control de calidad*: Se utiliza mucho en la industria y la estadística para analizar el comportamiento de un proceso productivo.

Para graficar un histograma se usa `plt.hist(X)` y como argumento se pasa la columna que quieras analizar. Además, se puede personalizar el gráfico con otros argumentos:

- `bins=n` → dice cuántas barras serán mostradas en el histograma. Por defecto vienen 10.A este argumento también se le puede pasar una lista para establecer un límite en el rango de las X en las se desea que se muestren.   
- Si se quiere colocar *2 histogramas superpuestos*, se puede usar el argumento `alpha` al que si se le pasa un valor entre 0 y 1, cambia la transparencia del color.
- También es posible usar el argumento `histtype='step'` que hace que cambie la forma que se representa el histograma (por ejemplo líneas en lugar de rectángulos sólidos). (Aquí figura la documentación)[https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.hist.html] para ver qué otros tipos de `histtype` existen así como otro tipo de argumentos de `hist`.
- Por último, si se usa el parámetro `density=True`, el gráfico se trasformará a porcentajes para que el área total de los rectángulos sea de 1 (100%). Cabe mencionar que este argumento porque podría ser muy útil para entender el *peso* de los valores más que su cantidad en sí misma.

```python
# Goles totales por equipo en 2018
df_2018 = df_football.query('year == 2018')
team_goals_2018 = df_2018.groupby('team_name')['goals'].sum()
```

```python

# Plot
plt.figure(figsize=[7, 6]) # 7 in de ancho y 6 in de alto
plt.hist(team_goals_2018, bins=25)
plt.title('Goles totales por equipo en 2018')
plt.xlabel('Cantidad de goles')
plt.ylabel('Número de equipos con X cantidad de goles')
plt.show()
```

Modificando el parámetro `density`:

```python
# Goles totales por equipo en 2018
df_2018 = df_football.query('year == 2018')
team_goals_2018 = df_2018.groupby('team_name')['goals'].sum()
plt.hist(team_goals_2018, bins=15, histtype='step', density=True)
plt.show()
```

### Gráfico de dispersión - (Scatter Plot)

Los **gráficos de dispersión**, también denominados *diagramas de dispersión* o *scatter plots*, se utilizan para comparar los valores que toman dos variables distintas, una de las cuales se representa a lo largo del *eje x* y la otra a lo largo del *eje y*.    

La gráfica resultante nos permite identificar visualmente la *posible correlación* entre las dos variables.


##### Características principales

• *Ejes*: La variable independiente o de control se sitúa en el eje horizontal (X) y la variable dependiente o medida en el eje vertical (Y).   
• *Puntos*: Cada punto en el plano representa un par de valores correspondientes a una observación individual.   
• *Correlación*: Permite identificar rápidamente si las variables están relacionadas de forma positiva, negativa o nula, así como detectar patrones o valores atípicos



Se genera con `plt.scatter(x, y)`. Adicionalmente, hay un par de argumentos muy útiles para pasar una tercera variable, por ejemplo, el *tiempo*.    
Con `c=col` se puede colorear cada punto con un degradado. Y con `s=col` es posible dar distintos tamaños a los puntos. Para ver otros argumentos de este tipo de gráfico, [aquí figura la documentación sobre ello](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.scatter.html#matplotlib.pyplot.scatter).

```python
plt.scatter(df_2018.shots, df_2018.goals, c=df_2018.games)
plt.title('Disparos vs. goles')
plt.xlabel('Disparos')
plt.ylabel('Goles')
plt.show()
```

### Gráfico circular (Pie Chart)

Los gráficos circulares son uno de los gráficos más populares para **visualizar proporciones**. 


##### Características principales

• *Relación parte-todo*: Muestra cómo se distribuye un total entre diferentes componentes.    
• *Número de categorías*: Se recomienda utilizar máximo 5 o 7 categorías para evitar saturación visual y facilitar la lectura.     
• *Expresión de datos*: Los valores se muestran habitualmente en porcentajes (%).


Para crear un gráfico circular con Matplotlib, se utiliza la función `plt.pie()`. En el siguiente ejemplo se utiliza, además de la lista de datos principal con códigos de cursos, una segunda lista con los nombres de las tecnologías impartidas en los mismos, que se incluirán en el gráfico como etiquetas mediante el parámetro `labels`.    

Como alternativa, ya se ha visto anteriormente, que es posible utilizar `plt.legend()` para añadir una leyenda independiente.

Recordar que es necesario ejecutar `plt.show()` al final del tramo de código para visualizar correctamente el gráfico.

```python
import pandas as pd
import matplotlib.pyplot as plt

cursos = [164, 127, 31, 31,12, 11]
etiquetas = ['Python','R','SQL','Power BI','Excel','ChatGPT']
diccionario = {'cursos':cursos, 'etiquetas':etiquetas}
python_pie_chart_df = pd.DataFrame(diccionario)

plt.pie(x = python_pie_chart_df.cursos, labels = python_pie_chart_df.etiquetas)
plt.show()
```

```python
plt.pie(x = python_pie_chart_df.cursos)
plt.legend(labels = python_pie_chart_df.etiquetas, loc = [0.95,0.35])
plt.show()
```

```python
# el parametro `autopct` indica que queremos que se muestre el porcentaje de cada porción del gráfico circular
plt.pie(x = python_pie_chart_df.cursos, labels = python_pie_chart_df.etiquetas, autopct='%1.1f%%') 
plt.show()
```

##### Personalizar el gráfico    

Los gráficos circulares en Matplotlib, así como los fragmentos que los componen, son altamente personalizables.    

Estas son algunas de las opciones que ofrece la función `pie.plot()`. 

Para cambiar el color de los framentos, se va a utilizar el parámetro `colors`. Este parámetro permite aplicar la paleta de colores que se desee al gráfico en cuestión. 



```python
paleta_personal = ['#03ef62','#06bdfc','#ff6ea9','#ff931e','#ff5400','#7933ff']
plt.pie(x = python_pie_chart_df.cursos, labels = python_pie_chart_df.etiquetas, colors = paleta_personal, autopct='%1.1f%%')
plt.show()
```

También es posible variar la distancia de las etiquetas y los porcentajes, utilizando los parámetros `labeldistance` y `pctdistance`, respectivamente, y elegir un ángulo para iniciar el corte con el parámetro `startangle`.

```python
plt.pie(x = python_pie_chart_df.cursos, labels = python_pie_chart_df.etiquetas, colors = paleta_personal, autopct='%1.1f%%', 
pctdistance= 0.8, labeldistance=1.1, startangle=180)

plt.show()
```

Por último, es posible destacar un fragmento extrayendo este mediante el parámetro `explode`, como se indica a continuación:

```python
explode = (0, 0.1, 0, 0, 0, 0)  # solo separamos ("explode") el segundo fragmento (en el ejemplo es 'R')

plt.pie(x = python_pie_chart_df.cursos, labels = python_pie_chart_df.etiquetas, 
        colors = paleta_personal, autopct='%1.1f%%', 
        pctdistance= 0.8, labeldistance=1.1, 
        startangle=180, explode=explode)

plt.show()
```

## Anotaciones sobre gráficos.

Usando `ax.annotate('notes', xy=(xcoor, ycoor), xytext=(xcoor, ycoor), arrowprops={'arrowstyle':'->', 'color':'gray'})` es posible poner texto sobre el gráfico en coordenadas específicas para recalcar algo.    

También es posible poner el texto sobre coordenadas específicas.    

Por último, se puede añadir una flecha que conecte el texto con el punto que se desea recalcar, y se puede personalizar la flecha dentro de un diccionario. 

```python
import numpy as np

fig, ax = plt.subplots()

t = np.arange(0.0, 5.0, 0.01)
s = np.cos(2*np.pi*t)
line, = ax.plot(t, s, lw=2)

ax.annotate('máximo local', xy=(2, 1), xytext=(3, 1.5),
            arrowprops=dict(facecolor='black', shrink=0.05),
            )

ax.set_ylim(-2, 2)  # limita el eje y entre -2 y 2
plt.show()
```
