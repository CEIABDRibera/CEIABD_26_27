# Apartado 2 · Vectores y matrices

**Objetivos:** representar características, identificar filas y columnas, consultar `shape` y combinar valores que ocupen posiciones correspondientes.


Un **vector** es una secuencia ordenada de valores. Una observación `[duración, distancia, pulsaciones]` solo se entiende si conocemos qué significa cada posición.    
Una **matriz** reúne observaciones en filas y características en columnas.

```python
registro = np.array([30, 4.2, 135])
print("Registro:", registro, "forma:", registro.shape)
semana_1 = np.array([2, 3, 4])
semana_2 = np.array([3, 2, 5])
print("Suma por posiciones:", semana_1 + semana_2)
print("Media semana 1:", np.mean(semana_1))
```

### Ejemplo resuelto: producto escalar

Horas `[2,3,4]` y precios por hora `[10,12,8]` dan un coste de $2·10+3·12+4·8=88$ €. Cada par de posiciones representa la misma actividad.

```python
horas = np.array([2, 3, 4])
coste_hora = np.array([10, 12, 8])
print("Coste total (€):", np.dot(horas, coste_hora))
```

### Matriz y ejes

`shape` devuelve `(filas, columnas)`. En `datos[:, 1]`, `:` selecciona todas las filas y `1` la segunda columna (Python cuenta desde cero). `axis=0` calcula una media por cada columna.

```python
datos = np.array([[2,3,4], [3,2,5], [4,4,3]])
print("Forma:", datos.shape)
print("Primera fila:", datos[0])
print("Segunda columna:", datos[:,1])
print("Medias por columna:", np.mean(datos, axis=0))
```

# Apartado 3 · Sistemas lineales y probabilidad

**Objetivos:** definir incógnitas, plantear y comprobar un sistema; distinguir probabilidad teórica y frecuencia observada.


**Ejemplo:** dos entradas A y una B cuestan 11 €; una A y dos B cuestan 10 €. Sea $x$ el precio de A e $y$ el de B: $2x+y=11$ y $x+2y=10$. Sustituyendo $y=11-2x$ se obtiene $x=4$, $y=3$. Comprobación: $2·4+3=11$ y $4+2·3=10$.

```python
A=np.array([[2,1],[1,2]])
b=np.array([11,10])
solucion=np.linalg.solve(A,b)
print("Precios A y B:", solucion)
print("Comprobación:", A @ solucion)
```

**Probabilidad teórica:** con 3 tarjetas azules y 2 rojas, $P(azul)=3/5=0,6$ (60 %), suponiendo tarjetas equiprobables.     
**Frecuencia observada:** si en diez extracciones se obtienen siete azules, es $7/10=0,7$.       
Son conceptos distintos; una muestra pequeña no tiene por qué repetir la proporción teórica.

```python
print("Probabilidad teórica:", 3/5)
print("Frecuencia observada:", 7/10)
```

# Apartado 4 · Estadística descriptiva

**Objetivos:** calcular e interpretar media, mediana, varianza y desviación típica; detectar el efecto de valores extremos.


Datos ficticios de tiempos, en minutos: `[10,12,12,14,17]`.    
**Media** $13$.    
**Mediana** $12$.      
La **varianza poblacional** es $(9+1+1+1+16)/5=5,6$ min² y la **desviación típica** $\sqrt{5,6}\approx2,37$ min.      

En este ejemplo usamos `np.var` y `np.std` con su configuración predeterminada (`ddof=0`).

```python
tiempos_ejemplo=np.array([10,12,12,14,17], dtype=float)
print("Media:",np.mean(tiempos_ejemplo))
print("Mediana:",np.median(tiempos_ejemplo))
print("Varianza poblacional:",np.var(tiempos_ejemplo))
print("Desviación típica poblacional:",np.std(tiempos_ejemplo))
```

Si sustituimos 17 por 37, la **media** cambia de 13 a 17, pero la **mediana** permanece en 12. No basta con decir que una medida «es mejor»: hay que explicar qué pregunta queremos responder.

```python
con_extremo=np.array([10,12,12,14,37])
print("Media y mediana con extremo:",np.mean(con_extremo),np.median(con_extremo))
```

# Apartado 5 · Gráficos y actividad integradora

**Objetivos:** elegir gráficos, etiquetar ejes, elaborar un análisis reproducible y reconocer sus límites.


Estos datos ficticios son **20 tiempos, en minutos**. El índice representa el número de observación; no presupongas que es una fecha.

```python
tiempos=np.array([12,15,14,13,16,18,17,15,14,16,19,18,17,16,15,20,19,18,17,16],dtype=float)
print("Registros:",len(tiempos))
```

### Ejemplo resuelto: gráfico de líneas

Un gráfico de líneas hace visible la secuencia de registros. Si el orden no corresponde a momentos de medida, no se puede presentar como evolución temporal.

```python
numero=np.arange(1,len(tiempos)+1)
plt.figure(figsize=(8,3))
plt.plot(numero,tiempos,marker="o")
plt.title("Tiempo registrado en cada observación")
plt.xlabel("Número de observación")
plt.ylabel("Tiempo (minutos)")
plt.grid(True)
plt.show()
```

### Ejemplo resuelto: barras para comparar grupos

Separar los diez primeros registros de los diez últimos es **una decisión didáctica del ejercicio**. No atribuyas los grupos a personas, turnos o fechas reales.

```python
media_1=np.mean(tiempos[:10]); media_2=np.mean(tiempos[10:])
plt.figure(figsize=(6,3))
plt.bar(["Registros 1–10","Registros 11–20"],[media_1,media_2])
plt.title("Comparación de tiempos medios")
plt.xlabel("Grupo de registros")
plt.ylabel("Tiempo medio (minutos)")
plt.show()
```
