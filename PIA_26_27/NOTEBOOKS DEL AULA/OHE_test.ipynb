# PANDAS -> getDummies()

```python
import pandas as pd

df = pd.DataFrame({
    "color": [
        "Rojo",
        "Azul",
        "Verde",
        "Rojo"
    ]
})

df_encoded = pd.get_dummies(
    df,
    columns=["color"]
)

print(df_encoded)
```
